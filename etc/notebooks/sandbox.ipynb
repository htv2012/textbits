# Sandbox

Here is the place to try out ideads

```python
import sqlite3
```

```python
conn = sqlite3.connect(':memory:')
```

```python
conn.execute('create table bits (id integer, name text, data text)')
```

```python
d = [
    (0, 'hello.py', 'print("hello")'),
    (1, 'bye.py', 'print("bye")'),
]
conn.executemany('insert into bits values (?, ?, ?)', d)
```

```python
list(
    conn.execute('select * from bits')
)
```

```python
for id_, name, content in conn.execute('select * from bits'):
    print(f"\n{id_}", end=" ")
    print(name)
    print(content)
```
